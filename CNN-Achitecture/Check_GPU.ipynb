```python
import torch

print("Check if CUDA is available: ", torch.cuda.is_available())
print("Check if CUDA is currently being used: ", torch.cuda.get_device_name())
```

```python
import torch.nn as nn
import torch.optim as optim
# Auto detect device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using:", device)
```

```python
class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc = nn.Linear(10, 1)

    def forward(self, x):
        return self.fc(x)
```

```python
# ✅ Move model to GPU
model = Net().to(device)
optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.MSELoss()
inputs = torch.randn(64, 10).to(device)   # ✅ Push data to GPU
targets = torch.randn(64, 1).to(device)   # ✅ Push labels to GPU

# Training step
model.train()
for epoch in range(10):
    optimizer.zero_grad()
    outputs = model(inputs)
    loss = criterion(outputs, targets)
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```
