```python
import torch
import torch.nn as nn
import torch.optim as optim

# AND gate data
X = torch.tensor([[0.,0.], [0.,1.], [1.,0.], [1.,1.]],requires_grad=True)
y = torch.tensor([[0.], [0.], [0.], [1.]],requires_grad=True)

class Perceptron(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(2, 1)

    def forward(self, x):
        z = self.fc(x)
        return torch.where(z > 0, 1.0, 0.0)  # Hard step activation

model = Perceptron()
optimizer = optim.SGD(model.parameters(), lr=0.1)


for epoch in range(50):
    optimizer.zero_grad()
    y_pred = model(X)
    loss = ((y_pred - y) ** 2).mean()  # MSE for step function
    loss.backward()
    optimizer.step()

print("Predictions:")
print(model(X))
```
