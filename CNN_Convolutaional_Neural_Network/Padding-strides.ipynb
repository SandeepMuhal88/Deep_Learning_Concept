```python
import tensorflow 
from tensorflow import keras
from keras import layers
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten
from keras.layers import Conv2D, MaxPooling2D
```

```python
from keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
model = Sequential([
    Conv2D(32,(3,3),activation='relu', padding='valid',input_shape=(28, 28, 1)),\
    Conv2D(32,(3,3),activation='relu', padding='valid'),
    Conv2D(64,(3,3),activation='relu', padding='valid'),

    Flatten(),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
]
)
```

```python
model.summary()
```

```python
model = Sequential([
    Conv2D(32,(3,3),activation='relu', padding='same',input_shape=(28, 28, 1)),\
    Conv2D(32,(3,3),activation='relu', padding='same'),
    Conv2D(64,(3,3),activation='relu', padding='same'),

    Flatten(),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
]
)
```

```python
model.summary()
```

### Strides

```python
model2 = Sequential([
    Conv2D(32,(3,3),activation='relu', padding='same',strides=(2,2),input_shape=(28, 28, 1)),\
    Conv2D(32,(3,3),activation='relu', padding='same', strides=(2,2)),
    Conv2D(64,(3,3),activation='relu', padding='same', strides=(2,2)),

    Flatten(),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
]
)
```
