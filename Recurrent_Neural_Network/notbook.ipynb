```python
import tensorflow
import numpy as np
import pandas as pd
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,SimpleRNN
```

```python
model=Sequential()
model.add(SimpleRNN(4,input_shape=(4,5)))
model.add(Dense(1,activation='relu'))
model.summary()
```

```python
print(model.get_weights()[0].shape)
model.get_weights()[0]
```
