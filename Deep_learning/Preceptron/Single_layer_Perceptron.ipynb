```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Make a Algorithm for single layer perceptron

# x1,x2 =inputs
# w1,w2 = weights
# b = bias
# y = output
# input-> 1layer-> stepfunction->output

def step(z):
    if z >=0:
        return 1
    else:
        return 0
```

```python
# We have a data in 2-D form
# calculate w1 and w2
w1=2.6
w2=1.5
b=0.1
# w1*x1+w2*x2+b

def predict(x1,x2):
    z=w1*x1+w2*x2+b
    k=step(z)
    return k



```
